# Question 2: Explain the difference between List and Tuple in Python.

**Answer:**

| List                                       | Tuple                                      |
| ------------------------------------------ | ------------------------------------------ |
| Lists are mutable.                         | Tuples are immutable.                      |
| Lists are written using `[ ]`.             | Tuples are written using `( )`.            |
| Elements can be added, removed or changed. | Elements cannot be changed after creation. |
| Example: `[1, 2, 3]`                       | Example: `(1, 2, 3)`                       |

**Example:**

```python
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)
```

#5. Write a Python program to check whether a number is even or odd using conditional statements.

**Solution:**
```
num = int(input("Enter a number: "))

if num % 2 == 0:
    print("The number is even.")
else:
    print("The number is odd.")

Example Output:

Enter a number: 8
The number is even.
```

#8. Write a Python program to remove duplicate values from a list using a set.

Solution:
```
numbers = [2, 3, 4, 2, 5, 2]

unique_numbers = set(numbers)

print("List after removing duplicates:", unique_numbers)
```
Output:

List after removing duplicates: {2, 3, 4, 5}

Note: A set automatically stores only unique values.

#9. Write a function that returns the square of a number.

**Solution:**
```
def square(num):
    return num * num

number = int(input("Enter a number: "))

print("Square:", square(number))
```
Example Output:

Enter a number: 5

Square: 25

#10. Write a Python program to count how many times a number appears in a list.Given list: [2, 3, 4, 2, 5, 2]

**Solution:**
```
numbers = [2, 3, 4, 2, 5, 2]

num = int(input("Enter the number to count: "))

count = numbers.count(num)

print("The number appears", count, "times.")
```
Example Output:

Enter the number to count: 2

The number appears 3 times.
